### Anatomia de uma Execução

#### Leitura Inicial (O Peso do CSV)

```python
# Célula de Setup
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
import time
```

```python
# Criando a sessão do Spark (o ponto de entrada)
spark = SparkSession.builder \
    .appName("Parallelism") \
    .master("local[*]") \
    .config("spark.ui.port", "4040") \
    .getOrCreate()

# Ajustando o nível de log para não poluir a tela
spark.sparkContext.setLogLevel("ERROR")

print("Spark iniciado com sucesso!")
```

#### Leitura dos CSVs

```python
print("--- Lendo arquivos CSV (Inferindo Schema) ---")

# Lendo Vendas
df_vendas = spark.read.csv("data/vendas.csv", header=True, inferSchema=True)

# Lendo Usuários
df_usuarios = spark.read.csv("data/usuarios.csv", header=True, inferSchema=True)

print("Leitura concluída.")

# Estrutura dos dados
df_vendas.printSchema()
df_usuarios.printSchema()
```

#### Particionamento

```python
qtd_part_vendas = df_vendas.rdd.getNumPartitions()
qtd_part_users = df_usuarios.rdd.getNumPartitions()

print(f"Partições de Vendas: {qtd_part_vendas}")
print(f"Partições de Usuários: {qtd_part_users}")
```

#### Salvando em Parquet

O Parquet já guarda o esquema (tipos de dados) e comprime os dados. Em csv o spark precisa ler linha a linha e inferir o schema

```python
df_vendas.write.mode("overwrite").parquet("data/vendas_parquet")
df_usuarios.write.mode("overwrite").parquet("data/usuarios_parquet")

print("Conversão concluída! Verifiquem a pasta 'work/dados' no menu à esquerda.")
```

#### Lazy Evaluation

```python
# Célula 5: Transformações (Lazy Evaluation)
print("--- Definindo Transformações (Plano de Execução) ---")
start_time = time.time()

# 1. Carregando dos Parquets (Leitura Otimizada)
vendas_parquet = spark.read.parquet("data/vendas_parquet")
users_parquet = spark.read.parquet("data/usuarios_parquet")

# 2. Transformação em VENDAS
vendas_proc = vendas_parquet.filter(F.col("valor") > 500) \
    .withColumn("valor_imposto", F.col("valor") * 0.10) \
    .withColumn("ano", F.year(F.from_unixtime("timestamp"))) \
    .select("id_venda", "id_user", "produto", "valor", "valor_imposto", "ano")

# 3. Transformação em USUÁRIOS
users_proc = users_parquet.filter(F.col("estado").isin("SP", "RJ")) \
    .withColumnRenamed("id_user", "id_cliente") \
    .select("id_cliente", "nome", "estado")

end_time = time.time()
print(f"Tempo de Definição: {end_time - start_time:.4f} segundos (Instantâneo!)")
```

#### Plano de Execução (DAG)

DAG – Directed Acyclic Graph - Grafo Acíclico Direcionado

    Tarefa 1 -> Tarefa 2 -> Tarefa 3 -> Tarefa 5
                         -> Tarefa 4 --^

    Tarefa 1 -> Tarefa 2 -> Tarefa 1 -> Tarefa 2 -> Tarefa 3

```python
# Vamos ver a "Receita" que o Spark criou
vendas_proc.explain()
```

```python
users_proc.explain()
```

#### A Ação (Action)

```python
print("--- Disparando o Processamento (Action) ---")
start_time = time.time()

# Disparando as duas contagens
qtd_vendas = vendas_proc.count()
qtd_users = users_proc.count()

end_time = time.time()

print(f"Vendas processadas: {qtd_vendas}")
print(f"Usuários processados (SP/RJ): {qtd_users}")
print(f"Tempo Total de Processamento: {end_time - start_time:.4f} segundos")
```
